```python
import numpy as np 
import pandas as pd 
```

```python
def load_3D_signals(subset):
    filepaths = [
        f"{subset}/Inertial Signals/body_acc_x_{subset}.txt",
        f"{subset}/Inertial Signals/body_acc_y_{subset}.txt",  
        f"{subset}/Inertial Signals/body_acc_z_{subset}.txt"
    ]

    loaded_signals = []
    for filepath in filepaths:
        data = pd.read_csv(filepath, sep='\s+', header=None)
        loaded_signals.append(data.values)

    return np.dstack(loaded_signals)

x_train = load_3D_signals('train')
y_test = load_3D_signals('test')
```
